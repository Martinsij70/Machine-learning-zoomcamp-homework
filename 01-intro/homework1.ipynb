```python
csv_asia7 = csv_asia[['model_year', 'vehicle_weight']].head(n=7)
csv_asia7
```

```python
X = csv_asia7.values
X
```

```python
XT = X.T
XT
```

```python
XTX = XT.dot(X)
XTX
```

```python
import numpy as np
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
v = XTX_inv.dot(XT)
v
```

```python
w = v.dot(y)
w
```
